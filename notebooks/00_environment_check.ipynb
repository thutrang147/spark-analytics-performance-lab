```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder    # khởi tạo cấu hình Spark Application
    .appName("SparkAnalyticsLab")   # đặt tên app
    .master("local[4]") # cho phép Spark thực thi bằng tối đa 4 luồng local
    .config("spark.driver.memory", "2g")    # cấu hình bộ nhớ driver
    .config("spark.sql.shuffle.partitions", "4")    # số partition mặc định cho nhiều phép shuffle của SparkSQL
    .getOrCreate()
)

sc = spark.sparkContext

print("Spark version:", spark.version)
print("Master:", sc.master)
print("Parallelism:", sc.defaultParallelism)
print("Spark UI:", sc.uiWebUrl)
```

```python
df = spark.range(0, 100, 1, 4)  # tạo dữ liệu từ 0 - 99, bước nhảy 1, chia thành 4 partitions

df.show(10)

print("Total rows:", df.count())
print("Partitions:", df.rdd.getNumPartitions())
```
